# Step 1: Load Data

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('OnlineRetail1.csv',encoding='latin')

secondData = data.copy()
data
```

# Step 2: Drop Missing Invoices or Descriptions

```python
data.info()
```

```python
data = data.dropna(how='any',subset=['Description'])
data
```

# Step 3: Remove Canceled Invoices <BR> Start with letter 'c'

```python
data = data[~data['InvoiceNo'].str.startswith('C')]
data
```

# Step 4: Keep Only Positive Quantities

```python
data = data[data['Quantity'] > 0]
data
```

# Step 5: Standardize Item Descriptions | Lower case and remove space

```python
data['Description'] = data['Description'].str.lower().str.replace(' ','')
data
```

# Step 6: Filter for a Single Country (e.g., United Kingdom)

```python
data[data['Country'] == 'United Kingdom']
```

# Step 7: Remove Duplicates

```python
data.drop_duplicates()
```

# Step 8: Add TotalPrice column , TotalPrice = Quantity *  UnitPrice

```python
secondData['TotalPrice'] = secondData['Quantity'] * secondData['UnitPrice']
secondData
```

# Step 11: Time-Based Analysis | Convert InvoiceDate to datetime and check popular months/hours.

```python
months = secondData['InvoiceDate'].str.split().str[0].str.split('/').str[0]

hours = secondData['InvoiceDate'].str.split().str[1].str.split(':').str[0]


months.plot(kind='bar',xlabel='Months')
plt.show()
```
